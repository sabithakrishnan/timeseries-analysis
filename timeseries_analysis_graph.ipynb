```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Statistical model
data = {
    "Model": ["ARIMA", "SARIMA", "ETS", "Prophet"],
    "R2 Score": [-0.0262,-1.1094,0.4954,0.3809],
    "Adjusted R2":[-0.0290,-1.1117,0.4915,0.3793],
    "RMSE": [1.4025,2.0108,0.4305,0.5281],
    "MAE":[1.0320,1.4818,0.4751,0.5432]
}
df = pd.DataFrame(data)

# Melt the dataframe to make it compatible with seaborn
df_melted = df.melt(id_vars="Model", var_name="Metric", value_name="Value")

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=df_melted, x="Metric", y="Value", hue="Model", palette="Set2")
plt.title("Model Comparison Across Performance Metrics")
plt.ylabel("Score / Error Value")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
#ML models
data2 = {
    "Model": ["LGBM", "XGBoost", "Random forest"],
    "R2 Score": [0.9433,0.8071,0.9243],
    "Adjusted R2":[0.9431,0.8065,0.9229],
    "RMSE": [2.0052,0.3680,0.0652],
    "MAE":[1.000,0.3960,0.1900]
}
df2 = pd.DataFrame(data2)

# Melt the dataframe to make it compatible with seaborn
df_melted2 = df2.melt(id_vars="Model", var_name="Metric", value_name="Value")

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=df_melted2, x="Metric", y="Value", hue="Model", palette="Set2")
plt.title("Model Comparison Across Performance Metrics")
plt.ylabel("Score / Error Value")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

#Neural network models
data3 = {
    "Model": ["CNN","LSTM","Transformer"],
    "R2 Score": [0.8092,0.8141,0.4961],
    "Adjusted R2":[0.8091,0.8139,0.4947],
    "RMSE": [2.8089,0.0482,0.6628],
    "MAE":[1.7172,0.0350,0.5073]
}
df3 = pd.DataFrame(data3)

# Melt the dataframe to make it compatible with seaborn
df_melted3 = df3.melt(id_vars="Model", var_name="Metric", value_name="Value")

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=df_melted3, x="Metric", y="Value", hue="Model", palette="Set2")
plt.title("Model Comparison Across Performance Metrics")
plt.ylabel("Score / Error Value")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import os
import cv2
import matplotlib.pyplot as plt
import numpy as np

# 1. Define the paths to your 10 images inside Colab
# Change these filenames to match your exact files in the Colab file explorer
image_files = [
    '/content/arima.png', '/content/sarima.png', '/content/ETS.png', '/content/prophet.png', '/content/lgbm.png',
    '/content/xgboost.png', '/content/random forest.png', '/content/cnn.png', '/content/lstm.png', '/content/transformer.png'
]

# 3. Set up the Matplotlib subplot grid (5 rows, 2 columns)
fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(15, 25))
axes = axes.flatten()  # Flatten the 5x2 array into a 1D list of 10 slots

# 4. Loop through and plot each image
for i, file_path in enumerate(image_files):
    # Read the image using OpenCV
    img = cv2.imread(file_path)

    # Convert BGR (OpenCV default) to RGB (Matplotlib default)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display the image in its grid slot
    axes[i].imshow(img_rgb)
    axes[i].set_title(f"Graph {i+1}", fontsize=14, fontweight='bold')
    axes[i].axis('off')  # Hides the pixel rulers/axes for a clean look

# 5. Clean up spacing and save the final single layout
plt.tight_layout()
plt.savefig('combined_forecast_grid.png', dpi=300, bbox_inches='tight')
plt.show()

print("Successfully generated and saved 'combined_forecast_grid.png' in your Colab files!")
```
